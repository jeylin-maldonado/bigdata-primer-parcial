## Ejercicio 1: Clasificación de edad

**Enunciado:** determinar si una persona es niño, adolescente, adulto o adulto mayor según su edad.

**Lógica:** se evalúa la edad con condiciones `if`, `elif` y `else`. El primer rango que se cumple determina la categoría.

```python
def clasificar_edad(edad):
    if edad < 13:
        return "Niño"
    elif edad <= 17:
        return "Adolescente"
    elif edad <= 59:
        return "Adulto"
    else:
        return "Adulto mayor"

edad = 20
print(f"Edad: {edad} años")
print(f"Categoría: {clasificar_edad(edad)}")
```

## Ejercicio 2: Conversor de temperatura

**Enunciado:** convertir una temperatura de grados Celsius a Fahrenheit.

**Lógica:** se aplica la fórmula `F = (C × 9/5) + 32` dentro de una función y se muestra el resultado.

```python
def convertir_a_fahrenheit(celsius):
    return (celsius * 9/5) + 32

celsius = 25
print("\nEJERCICIO 2")
print(f"Temperatura en Celsius: {celsius}°C")
print(f"Temperatura en Fahrenheit: {convertir_a_fahrenheit(celsius)}°F")
```

# EJERCICIO 3: Calculadora básica
# Enunciado: realizar suma, resta, multiplicación o división entre dos números.
# Lógica: se usa if, elif y else, validando que no se divida entre cero.

```python
def calculadora_basica(num1, num2, operacion):
    if operacion == "suma":
        return num1 + num2
    elif operacion == "resta":
        return num1 - num2
    elif operacion == "multiplicacion":
        return num1 * num2
    elif operacion == "division":
        if num2 == 0:
            return "No se puede dividir entre cero"
        return num1 / num2
    else:
        return "Operación no válida"

num1 = 20
num2 = 5
operacion = "division"

print("\nEJERCICIO 3")
print(f"Primer número: {num1}")
print(f"Segundo número: {num2}")
print(f"Operación: {operacion}")
print(f"Resultado: {calculadora_basica(num1, num2, operacion)}")
```

# EJERCICIO 4: Número par o impar
# Enunciado: determinar si un número entero es par o impar.
# Lógica: se usa el operador módulo %. Si el residuo entre 2 es 0, es par.

```python
def verificar_par_impar(numero):
    if numero % 2 == 0:
        return "Par"
    else:
        return "Impar"

numero = 8
print("\nEJERCICIO 4")
print(f"Número: {numero}")
print(f"Resultado: {verificar_par_impar(numero)}")
```

# EJERCICIO 5: Tabla de multiplicar
# Enunciado: mostrar la tabla de multiplicar del 1 al 12.
# Lógica: se usa un ciclo for para generar los resultados.


```python
def tabla_multiplicar(numero):
    for i in range(1, 13):
        print(f"{numero} x {i} = {numero * i}")

numero = 7
print("\nEJERCICIO 5")
print(f"Tabla de multiplicar del {numero}:")
tabla_multiplicar(numero)
```

# EJERCICIO 6: Contador de vocales
# Enunciado: contar cuántas vocales contiene una frase.
# Lógica: se convierte la frase a minúsculas y se revisa letra por letra.


```python
def contar_vocales(frase):
    vocales = "aeiou"
    contador = 0

    for letra in frase.lower():
        if letra in vocales:
            contador += 1

    return contador

frase = "La programacion en Python es muy util"

print("\nEJERCICIO 6")
print(f"Frase: {frase}")
print(f"Cantidad de vocales: {contar_vocales(frase)}")
```

# EJERCICIO 7: Promedio de calificaciones
# Enunciado: calcular suma, promedio e indicar si aprobó o reprobó.
# Lógica: se usa sum() para sumar y len() para dividir entre la cantidad de notas.

```python
calificaciones = [85, 90, 78, 69, 88]

suma_total = sum(calificaciones)
promedio = suma_total / len(calificaciones)

if promedio >= 70:
    estado = "Aprobado"
else:
    estado = "Reprobado"

print("\nEJERCICIO 7")
print(f"Calificaciones: {calificaciones}")
print(f"Suma total: {suma_total}")
print(f"Promedio: {promedio}")
print(f"Estado: {estado}")
```

# EJERCICIO 8: Lista de números positivos y negativos
# Enunciado: contar números positivos, negativos y ceros.
# Lógica: se recorre una lista y se compara cada número con cero.

```python
numeros = [10, -5, 0, 8, -3, 15, 0, -9, 4, -1]

positivos = 0
negativos = 0
ceros = 0

for numero in numeros:
    if numero > 0:
        positivos += 1
    elif numero < 0:
        negativos += 1
    else:
        ceros += 1

print("\nEJERCICIO 8")
print(f"Lista de números: {numeros}")
print(f"Cantidad de positivos: {positivos}")
print(f"Cantidad de negativos: {negativos}")
print(f"Cantidad de ceros: {ceros}")
```

# EJERCICIO 9: Mayor y menor de una lista
# Enunciado: encontrar el número mayor, menor y la diferencia entre ambos.
# Lógica: se usa max() para el mayor y min() para el menor.

```python
numeros = [45, 12, 89, 34, 7, 100, 56]

mayor = max(numeros)
menor = min(numeros)
diferencia = mayor - menor

print("\nEJERCICIO 9")
print(f"Lista de números: {numeros}")
print(f"Número mayor: {mayor}")
print(f"Número menor: {menor}")
print(f"Diferencia entre mayor y menor: {diferencia}")
```

# EJERCICIO 11: Función para calcular descuento
# Enunciado: calcular el precio final aplicando un porcentaje de descuento.
# Lógica: se calcula el descuento y se resta del precio original.

```python
def calcular_descuento(precio_original, porcentaje_descuento):
    descuento = precio_original * (porcentaje_descuento / 100)
    precio_final = precio_original - descuento
    return precio_final

precio_original = 1500
porcentaje_descuento = 20

print("\nEJERCICIO 11")
print(f"Precio original: L. {precio_original}")
print(f"Porcentaje de descuento: {porcentaje_descuento}%")
print(f"Precio final: L. {calcular_descuento(precio_original, porcentaje_descuento)}")
```

# EJERCICIO 12: Validación de contraseña
# Enunciado: validar que una contraseña tenga 8 caracteres, número, mayúscula y minúscula.
# Lógica: se usan len(), isdigit(), isupper() e islower().

```python
def validar_contrasena(contrasena):
    tiene_8_caracteres = len(contrasena) >= 8
    tiene_numero = any(caracter.isdigit() for caracter in contrasena)
    tiene_mayuscula = any(caracter.isupper() for caracter in contrasena)
    tiene_minuscula = any(caracter.islower() for caracter in contrasena)

    if tiene_8_caracteres and tiene_numero and tiene_mayuscula and tiene_minuscula:
        return "Contraseña válida"
    else:
        return "Contraseña no válida"

contrasena = "Python2026"

print("\nEJERCICIO 12")
print(f"Contraseña: {contrasena}")
print(f"Resultado: {validar_contrasena(contrasena)}")
```

# EJERCICIO 13: Sistema simple de inventario
# Enunciado: crear productos, buscar por código y calcular valor total.
# Lógica: se usa una lista de diccionarios y se multiplica precio por cantidad.


```python
productos = [
    {"codigo": "P001", "nombre": "Teclado", "precio": 350, "cantidad": 10},
    {"codigo": "P002", "nombre": "Mouse", "precio": 180, "cantidad": 15},
    {"codigo": "P003", "nombre": "Monitor", "precio": 3200, "cantidad": 5}
]

print("\nEJERCICIO 13")
print("Lista de productos:")

for producto in productos:
    print(producto)

codigo_buscar = "P002"
producto_encontrado = None

for producto in productos:
    if producto["codigo"] == codigo_buscar:
        producto_encontrado = producto

print("\nProducto buscado:")
print(producto_encontrado)

valor_total = 0

for producto in productos:
    valor_total += producto["precio"] * producto["cantidad"]

print(f"Valor total del inventario: L. {valor_total}")
```

# EJERCICIO 14: Registro de ventas
# Enunciado: calcular total por venta, total general y compra más alta.
# Lógica: se multiplica cantidad por precio unitario y se usa max() para la venta mayor.

```python
ventas = [
    {"cliente": "Ana", "producto": "Laptop", "cantidad": 1, "precio_unitario": 15000},
    {"cliente": "Carlos", "producto": "Mouse", "cantidad": 3, "precio_unitario": 180},
    {"cliente": "María", "producto": "Monitor", "cantidad": 2, "precio_unitario": 3200}
]

total_general = 0

print("\nEJERCICIO 14")
print("Registro de ventas:")

for venta in ventas:
    venta["total"] = venta["cantidad"] * venta["precio_unitario"]
    total_general += venta["total"]
    print(f"Cliente: {venta['cliente']}, Producto: {venta['producto']}, Total: L. {venta['total']}")

compra_mas_alta = max(ventas, key=lambda venta: venta["total"])

print(f"Total general vendido: L. {total_general}")
print(f"Cliente con la compra más alta: {compra_mas_alta['cliente']}")
print(f"Monto de la compra más alta: L. {compra_mas_alta['total']}")
```

# EJERCICIO 15: Limpieza básica de texto
# Enunciado: limpiar una frase, convertirla a minúsculas y contar palabras.
# Lógica: se usan strip(), lower(), replace() y split().


```python
frase = "   Python   es   muy   importante para Big Data   "

frase_limpia = frase.strip()
frase_limpia = frase_limpia.lower()

while "  " in frase_limpia:
    frase_limpia = frase_limpia.replace("  ", " ")

palabras = frase_limpia.split()
cantidad_palabras = len(palabras)

print("\nEJERCICIO 15")
print(f"Frase original: '{frase}'")
print(f"Frase limpia: '{frase_limpia}'")
print(f"Cantidad de palabras: {cantidad_palabras}")
```
