# EJERCICIO 16: DataFrame de estudiantes
# Enunciado: crear un DataFrame con al menos 10 estudiantes, incluyendo nombre, carrera, edad, promedio y ciudad.
# Lógica: se usa pandas para crear una tabla de datos. Luego se muestran las primeras filas, información general,
# promedio general, estudiantes con promedio mayor o igual a 80 y ordenamiento por promedio.

```python
import pandas as pd

datos_estudiantes = {
    "nombre": ["Ana", "Carlos", "María", "Luis", "Sofía", "Pedro", "Lucía", "Juan", "Valeria", "Diego"],
    "carrera": ["Computación", "Administración", "Computación", "Derecho", "Big Data", "Computación", "Big Data", "Administración", "Computación", "Derecho"],
    "edad": [20, 22, 19, 21, 23, 20, 24, 22, 21, 25],
    "promedio": [85, 78, 92, 70, 88, 65, 95, 80, 89, 74],
    "ciudad": ["Tegucigalpa", "San Pedro Sula", "Choluteca", "La Ceiba", "Tegucigalpa", "Comayagua", "Valle", "Danlí", "Tegucigalpa", "Santa Rosa"]
}

df_estudiantes = pd.DataFrame(datos_estudiantes)

print("Primeras 5 filas:")
print(df_estudiantes.head())

print("\nInformación general del DataFrame:")
df_estudiantes.info()

print("\nPromedio general de calificaciones:")
print(df_estudiantes["promedio"].mean())

print("\nEstudiantes con promedio mayor o igual a 80:")
print(df_estudiantes[df_estudiantes["promedio"] >= 80])

print("\nEstudiantes ordenados por promedio de mayor a menor:")
print(df_estudiantes.sort_values(by="promedio", ascending=False))
```

# EJERCICIO 17: Análisis de ventas con pandas
# Enunciado: crear un DataFrame de ventas con al menos 15 registros, incluyendo fecha, vendedor, producto,
# categoría, cantidad y precio unitario.
# Lógica: se crea una columna TotalVenta multiplicando cantidad por precio unitario. Luego se agrupan datos
# para calcular ventas por vendedor, producto más vendido, categoría con mayor ingreso y ordenamiento.

```python
import pandas as pd

datos_ventas = {
    "fecha": ["2026-05-01", "2026-05-01", "2026-05-02", "2026-05-02", "2026-05-03",
              "2026-05-03", "2026-05-04", "2026-05-04", "2026-05-05", "2026-05-05",
              "2026-05-06", "2026-05-06", "2026-05-07", "2026-05-07", "2026-05-08"],
    "vendedor": ["Ana", "Carlos", "Ana", "María", "Carlos", "Luis", "Ana", "María", "Luis", "Carlos",
                 "Ana", "María", "Luis", "Carlos", "Ana"],
    "producto": ["Laptop", "Mouse", "Teclado", "Monitor", "Mouse", "Laptop", "Teclado", "Monitor", "Mouse", "Laptop",
                 "Monitor", "Teclado", "Laptop", "Mouse", "Monitor"],
    "categoria": ["Tecnología", "Accesorios", "Accesorios", "Tecnología", "Accesorios", "Tecnología", "Accesorios",
                  "Tecnología", "Accesorios", "Tecnología", "Tecnología", "Accesorios", "Tecnología", "Accesorios", "Tecnología"],
    "cantidad": [1, 3, 2, 1, 5, 1, 4, 2, 6, 1, 1, 3, 2, 4, 1],
    "precio_unitario": [15000, 180, 350, 3200, 180, 15000, 350, 3200, 180, 15000, 3200, 350, 15000, 180, 3200]
}

df_ventas = pd.DataFrame(datos_ventas)

df_ventas["TotalVenta"] = df_ventas["cantidad"] * df_ventas["precio_unitario"]

print("DataFrame de ventas:")
print(df_ventas)

print("\nTotal vendido por vendedor:")
print(df_ventas.groupby("vendedor")["TotalVenta"].sum())

producto_mas_vendido = df_ventas.groupby("producto")["cantidad"].sum().idxmax()
print(f"\nProducto más vendido: {producto_mas_vendido}")

categoria_mayor_ingreso = df_ventas.groupby("categoria")["TotalVenta"].sum().idxmax()
print(f"Categoría con mayor ingreso: {categoria_mayor_ingreso}")

print("\nVentas ordenadas de mayor a menor:")
print(df_ventas.sort_values(by="TotalVenta", ascending=False))
```

# EJERCICIO 18: Limpieza de datos con pandas
# Enunciado: crear un DataFrame con datos incompletos de clientes, identificar valores nulos,
# contar valores nulos por columna, reemplazar ciudades vacías y eliminar registros sin correo.
# Lógica: se usan isnull(), sum(), fillna() y dropna() para detectar, contar, reemplazar y eliminar datos incompletos.


```python
import pandas as pd

datos_clientes = {
    "nombre": ["Ana", "Carlos", "María", "Luis", "Sofía", "Pedro"],
    "edad": [25, None, 30, 22, None, 28],
    "correo": ["ana@gmail.com", None, "maria@gmail.com", "luis@gmail.com", None, "pedro@gmail.com"],
    "ciudad": ["Tegucigalpa", None, "San Pedro Sula", None, "Choluteca", "Valle"],
    "telefono": ["9999-1111", "9999-2222", None, "9999-4444", "9999-5555", None]
}

df_clientes = pd.DataFrame(datos_clientes)

print("DataFrame original:")
print(df_clientes)

print("\nValores nulos identificados:")
print(df_clientes.isnull())

print("\nCantidad de valores nulos por columna:")
print(df_clientes.isnull().sum())

df_clientes["ciudad"] = df_clientes["ciudad"].fillna("No especificada")
df_clientes_limpio = df_clientes.dropna(subset=["correo"])

print("\nDataFrame limpio:")
print(df_clientes_limpio)
```

# EJERCICIO 19: Análisis de préstamos
# Enunciado: crear un DataFrame con solicitudes de préstamo, calcular monto promedio, contar solicitudes por estado,
# mostrar préstamos aprobados, clientes con monto mayor a 50,000 y agrupar por estado.
# Lógica: se usan mean(), value_counts(), filtros condicionales y groupby() para analizar los préstamos.

```python
import pandas as pd

datos_prestamos = {
    "cliente": ["Ana", "Carlos", "María", "Luis", "Sofía", "Pedro", "Lucía", "Juan"],
    "monto_solicitado": [30000, 75000, 50000, 120000, 45000, 60000, 90000, 25000],
    "plazo": [12, 24, 18, 36, 12, 24, 36, 6],
    "ingreso_mensual": [15000, 25000, 18000, 40000, 17000, 22000, 35000, 12000],
    "estado_solicitud": ["Aprobado", "Aprobado", "Pendiente", "Rechazado", "Aprobado", "Pendiente", "Aprobado", "Rechazado"]
}

df_prestamos = pd.DataFrame(datos_prestamos)

print("DataFrame de préstamos:")
print(df_prestamos)

print("\nMonto promedio solicitado:")
print(df_prestamos["monto_solicitado"].mean())

print("\nSolicitudes por estado:")
print(df_prestamos["estado_solicitud"].value_counts())

print("\nPréstamos aprobados:")
print(df_prestamos[df_prestamos["estado_solicitud"] == "Aprobado"])

print("\nClientes con monto solicitado mayor a 50,000:")
print(df_prestamos[df_prestamos["monto_solicitado"] > 50000])

print("\nPromedio del monto solicitado por estado:")
print(df_prestamos.groupby("estado_solicitud")["monto_solicitado"].mean())
```

# EJERCICIO 20: Mini análisis de datos tipo empresa
# Enunciado: crear un DataFrame con empleados, mostrar primeros registros, estadísticas de salarios,
# salario promedio por departamento, empleado con salario más alto, empleados con más de 5 años,
# ordenar por salario y crear NivelExperiencia.
# Lógica: se usan head(), describe(), groupby(), idxmax(), filtros, sort_values() y apply()
# para analizar la información de empleados.

```python
import pandas as pd

datos_empleados = {
    "nombre": ["Ana", "Carlos", "María", "Luis", "Sofía", "Pedro", "Lucía", "Juan", "Valeria", "Diego"],
    "departamento": ["TI", "Ventas", "TI", "Contabilidad", "Ventas", "TI", "RRHH", "Contabilidad", "TI", "RRHH"],
    "cargo": ["Analista", "Vendedor", "Programadora", "Contador", "Supervisora", "Soporte", "Asistente", "Auditor", "Desarrolladora", "Coordinador"],
    "salario": [22000, 18000, 30000, 25000, 28000, 16000, 15000, 27000, 35000, 26000],
    "anios_experiencia": [3, 2, 6, 5, 7, 1, 2, 8, 9, 6],
    "ciudad": ["Tegucigalpa", "San Pedro Sula", "Tegucigalpa", "Choluteca", "La Ceiba", "Comayagua", "Valle", "Danlí", "Tegucigalpa", "Santa Rosa"]
}

df_empleados = pd.DataFrame(datos_empleados)

print("Primeros registros:")
print(df_empleados.head())

print("\nEstadísticas generales de salarios:")
print(df_empleados["salario"].describe())

print("\nSalario promedio por departamento:")
print(df_empleados.groupby("departamento")["salario"].mean())

indice_salario_alto = df_empleados["salario"].idxmax()
empleado_salario_alto = df_empleados.loc[indice_salario_alto]

print("\nEmpleado con salario más alto:")
print(empleado_salario_alto)

print("\nEmpleados con más de 5 años de experiencia:")
print(df_empleados[df_empleados["anios_experiencia"] > 5])

print("\nEmpleados ordenados por salario:")
print(df_empleados.sort_values(by="salario", ascending=False))

def clasificar_experiencia(anios):
    if anios <= 2:
        return "Junior"
    elif anios <= 5:
        return "Intermedio"
    else:
        return "Senior"

df_empleados["NivelExperiencia"] = df_empleados["anios_experiencia"].apply(clasificar_experiencia)

print("\nDataFrame con columna NivelExperiencia:")
print(df_empleados)
```
